```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

T = 16
t = np.arange(T + 1)

eps = 0.03
mse = eps + (1 - eps) * (1 - t / 15) ** 2
mse = np.maximum(mse, eps)

# Key times
t0, t1, t2 = 2, 7, 15

fig, ax = plt.subplots(figsize=(10, 6))

# 4DVarNet MSE trajectory
ax.plot(
    t, mse,
    marker="o",
    linewidth=2.5,
    color="tab:blue",
    markerfacecolor="cornflowerblue",
    markeredgecolor="tab:blue",
    label="4DVarNet: final MSE supervision"
)

# Vertical guides
for tx in [t0, t1, t2]:
    ax.axvline(tx, color="gray", linestyle="--", alpha=0.45)

# Labels t, t', t''
ax.text(t0 + 0.05, mse[t0] + 0.05, r"$t$", fontsize=18, fontweight="bold")
ax.text(t1 + 0.05, mse[t1] + 0.05, r"$t'$", fontsize=18, fontweight="bold")
ax.text(t2 + 0.05, mse[t2] + 0.05, r"$t''$", fontsize=18, fontweight="bold")

# Orange: pairwise direct transition t -> t''
ax.annotate(
    "",
    xy=(t2, mse[t2] + 0.02),
    xytext=(t0, mse[t0] + 0.02),
    arrowprops=dict(
        arrowstyle="-|>",
        lw=3,
        color="tab:orange",
        mutation_scale=22,
        connectionstyle="arc3,rad=-0.18"
    ),
)
ax.text(
    7.5, 0.68,
    r"Pairwise consistency:" "\n" r"direct transition $g(t,t'')$",
    color="tab:orange",
    fontsize=12
)

# Green: self-consistency composition t -> t' then t' -> t''
ax.annotate(
    "",
    xy=(t1, mse[t1] + 0.02),
    xytext=(t0, mse[t0] + 0.02),
    arrowprops=dict(
        arrowstyle="-|>",
        lw=3,
        color="green",
        mutation_scale=22,
        connectionstyle="arc3,rad=-0.18"
    ),
)
ax.annotate(
    "",
    xy=(t2, mse[t2] + 0.02),
    xytext=(t1, mse[t1] + 0.02),
    arrowprops=dict(
        arrowstyle="-|>",
        lw=3,
        color="green",
        mutation_scale=22,
        connectionstyle="arc3,rad=-0.18"    ),
)

ax.text(
    6.5, 0.48,
    r"Self-consistency:" "\n" r"first step $g(t,t')$",
    color="green",
    fontsize=12
)
ax.text(
    10.0, 0.15,
    r"Self-consistency:" "\n" r"second step $g(t',t'')$",
    color="green",
    fontsize=12
)

# Blue explanatory annotation
ax.text(
    0.9, 0.92,
    "4DVarNet trained by\n"
    "minimizing the final MSE\n"
    r"at $t=T$ (end of trajectory)",
    color="tab:blue",
    fontsize=12
)

# Legend
legend_elements = [
    Line2D([0], [0], color="tab:blue", marker="o", lw=2.5,
           label="4DVarNet: final MSE supervision"),
    Line2D([0], [0], color="tab:orange", lw=3,
           label=r"Pairwise consistency $(t \rightarrow t'')$"),
    Line2D([0], [0], color="green", lw=3,
           label=r"Self-consistency by composition $(t \rightarrow t',\, t' \rightarrow t'')$")
]
ax.legend(handles=legend_elements, loc="upper right", frameon=True)

# Formatting
ax.set_xlabel(r"Solver / diffusion time $t$", fontsize=14)
ax.set_ylabel("Optimization trajectory", fontsize=14)
ax.set_title(
    "4DVarNet supervision vs pairwise and self-consistency training",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlim(0, 16.3)
ax.set_ylim(0, 1.08)
ax.set_xticks(np.arange(0, 17, 1))
ax.grid(True, alpha=0.25)

plt.tight_layout()
plt.show()
```
